```python
import matplotlib.pyplot as plt
import numpy as np
from google.colab import drive
import os
# Data
assets = ['ASII.JK', 'MYOR.JK', 'BBNI.JK', 'PTBA.JK']
numerical_results = [1, 0, 1, 0]
analytic_results = [1, 1, 0, 0]

# Create figure and axis
fig, ax = plt.subplots(figsize=(3.5, 3))

# Set the width of each bar and positions of the bars
width = 0.35
x = np.arange(len(assets))

# Create bars
rects1 = ax.bar(x - width/2, numerical_results, width, label='Optimized Ansatz', color='#4285F4', edgecolor='black')
rects2 = ax.bar(x + width/2, analytic_results, width, label='Exact Solution', color='#34A853', edgecolor='black')

# Customize the plot
ax.set_ylabel('Allocation', fontsize=8)
ax.set_xlabel('Assets', fontsize=8)
ax.set_title('Optimized Ansatz vs. Exact Solution', fontsize=10, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(assets, fontsize=8)
ax.set_yticks([0, 1])
ax.tick_params(axis='both', which='major', labelsize=8)

# Add value labels on top of each bar
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=8)

autolabel(rects1)
autolabel(rects2)

# Add gridlines
ax.grid(axis='y', linestyle='--', alpha=0.3)

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Adjust layout and add legend below the plot
plt.tight_layout()
plt.legend(fontsize=8, loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2)

# Mount the Google Drive
drive.mount('/content/drive')

# Get the directory of the current notebook
notebook_dir = os.path.dirname('/content/drive/My Drive/Colab Notebooks/Kevin_plot.ipynb')

# Save the figure in the same directory as the notebook
plt.savefig(os.path.join(notebook_dir, 'xxxxx.png'), dpi=300, bbox_inches='tight')




# Show the plot
plt.show()
```
